## Configuration
Choose discovery or an explicit `tasks` list; `task_options` selects per-task checkpoint hashes and display names. Exports are separate per task. Loaded/unloaded cohorts and profiles stay separate. Missing requested tasks raise an error; incomplete recordings are excluded. Ambiguous checkpoints are never auto-selected.


```python
from b1z1_paper_names import DISPLAY_NAMES, LATEX_METHOD_NAMES
# Paper labels differ from implementation ablation numbers; task IDs stay unchanged.
CONFIG = {
    "input_roots": [str(ROOT / "evaluation/exp2")],  # Recursively inspect this experiment folder.
    "task": None,  # Legacy single-task mode uses output_dir directly.
    "tasks": None,  # Or ["b1z1_pact", "b1z1_unifp_original", "b1z1_ppo_pos"].
    "display_names": dict(DISPLAY_NAMES),
    "task_options": {},  # Per task: {"checkpoint_sha256": "...", "display_name": "..."}.
    "checkpoint_sha256": None,
    "expected_seeds": [0, 1, 2],
    "loading": ["loaded", "unloaded"],
    "output_dir": str(ROOT / "analysis/exp2"),  # Batch output: <output_dir>/<task>/.
    "histogram_edges": [i / 100 for i in range(-50, 51)],
    "acceleration_metrics": False,
}
# Optional JSON overrides support unattended execution without editing this notebook.
if os.environ.get("EXP2_ANALYSIS_CONFIG"):
    CONFIG.update(json.loads(os.environ["EXP2_ANALYSIS_CONFIG"]))
```

```python
from experiment_batch import plan_summaries
plans, inventory = plan_summaries(CONFIG, experiment=2)
display(pd.DataFrame(inventory))  # Tasks, checkpoints, seeds and output destinations.
summaries = {}
for index, settings in enumerate(plans, 1):
    print(f"[{index}/{len(plans)}] {settings['task']}")
    summary = analysis.summarize(settings)  # Preserve strict per-method validation/statistics.
    summaries[settings["task"]] = summary
    display(pd.DataFrame(summary["aggregates"]))
    print("\n".join(summary["diagnostics"]) or "No missing-data diagnostics.")
    print("Exported:", settings["output_dir"])
```
